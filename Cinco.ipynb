```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Ajustar máximo de filas
pd.options.display.max_rows = None

# 1. Corroboramos valores nulos
print("--- Valores Nulos ---")
print(df.isnull().sum())

# 2. Número total de datos válidos (n)
n = df['price'].count()

# 3. Límites superior e inferior
Min = df['price'].min()
Max = df['price'].max()
Limites = [Min, Max]

# 4. Rango (R)
R = Max - Min

# 5. Regla de Sturges para número de intervalos (ni)
ni = 1 + 3.322 * np.log10(n)
ni_redondeado = int(round(ni))

# 6. Ancho del intervalo (i)
i = R / ni_redondeado

# Mostrar resultados
print("\n--- Resultados del Análisis de Frecuencias ---")
print(f"Total datos (n): {n}")
print(f"Límites [Min, Max]: {Limites}")
print(f"Rango (R): {R:.2f}")
print(f"Intervalos Sturges (ni exacto): {ni:.2f} -> Redondeado: {ni_redondeado}")
print(f"Ancho del intervalo (i): {i:.2f}")

# 7. Crear los intervalos dinámicamente con Min y Max reales
intervalos = np.linspace(Min, Max, ni_redondeado + 1)
print("\n--- Intervalos Generados ---")
display(intervalos)

# 8. Generar las etiquetas de categorías automáticamente con los límites exactos
categorias = [f"Cat {idx+1} ({intervalos[idx]:.2f} - {intervalos[idx+1]:.2f})" for idx in range(len(intervalos)-1)]

# 9. Asignar las categorías en una nueva columna (para no perder el precio numérico original)
df['price_categoria'] = pd.cut(x=df['price'], bins=intervalos, labels=categorias, include_lowest=True)

# 10. Tabla de frecuencias
Tabla_freq = df['price_categoria'].value_counts().sort_index().reset_index()
Tabla_freq.columns = ['price_categoria', 'count']
print("\n--- Tabla de Frecuencias ---")
display(Tabla_freq)

# 11. Filtrar categorías con frecuencia > 0
Filtro = Tabla_freq[Tabla_freq['count'] > 0]
Filtro_index = Filtro.set_index('price_categoria')

# 12. Gráfico de barras
Filtro_index.plot(kind='bar', width=0.8, figsize=(11, 5), color="royalblue", rot=45, legend=False)
plt.title('Distribución de Frecuencias por Categorías de Precio')
plt.xlabel('Rango de Categorías de Precio ($)')
plt.ylabel('Frecuencia (Cantidad de Alojamientos)')
plt.tight_layout()
plt.show()
```
